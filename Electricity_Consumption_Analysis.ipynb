```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Read external dataset
df = pd.read_csv(
    "electricity.csv",
    parse_dates=["Date"]
)

# Handle missing values
df.loc[4, "kWh"] = np.nan

df["kWh"] = df["kWh"].fillna(
    df["kWh"].median()
)

# Plot electricity consumption
plt.figure(figsize=(12, 6))

plt.plot(
    df["Date"],
    df["kWh"],
    marker="o",
    linestyle="--",
    label="kWh"
)

plt.xlabel("Month")
plt.ylabel("Consumption (kWh)")

plt.legend()

plt.xticks(rotation=45)

plt.tight_layout()

# Save output
plt.savefig(
    "electricity_consumption_trend.png",
    dpi=300
)

plt.show()
```
